
feature engineering for BodyFat.csv (extended dataset)

adds a 'NavyBodyFat' column using the US Navy circumference method,
and drops the 'Original' y/n column.

US Navy formula (imperial units: inches):
    Men:   %BF = 86.010 * log10(Abdomen - Neck) - 70.041 * log10(Height) + 36.76
    Women: %BF = 163.205 * log10(Waist + Hip - Neck) - 97.684 * log10(Height) - 78.387


```python
import numpy as np
import pandas as pd

df = pd.read_csv("../data/raw/BodyFat.csv")

df = df.drop(columns="Original")

df["Height"] = df["Height"] *100 # m to cm
df["Height_in"] = df["Height"] / 2.54
df["Abdomen_in"] = df["Abdomen"] / 2.54
df["Neck_in"] = df["Neck"] / 2.54
df["Hip_in"] = df["Hip"] / 2.54 


def navy_bodyfat_male(abdomen, neck, height):
    return 86.010 * np.log10(abdomen - neck) - 70.041 * np.log10(height) + 36.76


def navy_bodyfat_female(waist, hip, neck, height):
    return 163.205 * np.log10(waist + hip - neck) - 97.684 * np.log10(height) - 78.387


def compute_navy(row):
    is_male = str(row["Sex"]).strip().upper()=="M"
    if is_male:
        return navy_bodyfat_male(row["Abdomen_in"], row["Neck_in"], row["Height_in"])
    else:
        return navy_bodyfat_female(row["Abdomen_in"], row["Hip_in"], row["Neck_in"], row["Height_in"])



df["NavyBodyFat"] = df.apply(compute_navy, axis=1)
```


Feature usefulness check for BodyFat_engineered.csv

1. Correlation of every numeric feature with BodyFat (target) - all rows
2. Same, split by Sex (M / F) separately

This is a quick screening step - high correlation = likely useful,
low correlation overall but high within one sex = sex-specific signal.


```python


EXCLUDE = ["BodyFat", "Density", "NavyBodyFat", "Sex"]

feature_cols = [c for c in df.columns if c not in EXCLUDE and pd.api.types.is_numeric_dtype(df[c])]


def correlation_report(data: pd.DataFrame):
    
    corrs = data[feature_cols].corrwith(data["BodyFat"]).sort_values(key=abs, ascending=False)
    for feat, val in corrs.items():
        print(f"  {feat}         {val:+.3f}")

print("all rows")
correlation_report(df)


```

```python
print("male")
male = df[df["Sex"].str.strip().str.upper() == "M"]
correlation_report(male)

print("\nfemale")
female = df[df["Sex"].str.strip().str.upper() == "F"]
correlation_report(female)
```

The "all rows" correlations are misleadingly weak compared to per-sex. Abdomen jumps from 0.357 (all) to 0.81/0.74 split.

This alone is strong evidence we should train separate models per sex.


1. Age is the standout sex-difference feature. Positive for males (+0.29, older men tend toward more fat) but essentially zero/slightly negative for females (-0.06).
2. Forearm, Biceps, Wrist, Ankle, Knee are all more predictive for females than males.
3. Height is essentially useless in both groups (-0.09, +0.01).
4. Chest is much more informative for men (0.70) than women (0.44).

### Anomaly & Outlier Removal

Anthropometric and body composition datasets frequently contain data entry errors, measurement inconsistencies, or extreme physical outliers (such as incorrect height/weight records or corrupted density measurements). 

Because regression models and linear scaling algorithms are highly sensitive to extreme values—especially on small to medium-sized sample sizes—leaving these anomalies in place skews feature weights and degrades performance. 

Applying an Interquartile Range (IQR) filter across the physical measurement features trims these extreme data points and stabilizes model training.

```python
numeric_cols = [
    "Age", "Weight", "Height", "Neck", "Chest", 
    "Abdomen", "Hip", "Thigh", "Knee", "Ankle", 
    "Biceps", "Forearm", "Wrist"
]

for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    df = df[(df[col] >= lower_bound) & (df[col] <= upper_bound)]

df = df.reset_index(drop=True)
```

### Dataset Segmentation and Export

Based on the correlation analysis, I separated the dataset into male and female cohorts. Biological fat distribution differs fundamentally between sexes, which is directly reflected in the feature correlations. For example, age shows a positive correlation for males but is negligible for females, while extremity measurements like biceps and forearms carry stronger predictive weight for females. 

Training a single unified model would force the algorithm to compromise between these conflicting patterns, leading to degraded performance. By splitting the data and manually isolating the most predictive, non-redundant features for each sex, the resulting models will be far more accurate and less prone to overfitting. 

This final step subsets the selected features, splits the dataframe, and exports the clean, preprocessed datasets to be ingested by the model training script.

```python
import os

FEATURES_BY_SEX = {
    "Male":   ["Abdomen", "Chest", "Hip", "Weight", "Thigh", "Neck"],
    "Female": ["Abdomen", "Hip", "Weight", "Biceps", "Thigh", "Forearm"],
}

datasets = {
    "Male":   df[df["Sex"].str.strip().str.upper() == "M"].reset_index(drop=True),
    "Female": df[df["Sex"].str.strip().str.upper() == "F"].reset_index(drop=True),
}

os.makedirs("../data/processed", exist_ok=True)

male_cols = FEATURES_BY_SEX["Male"] + ["BodyFat"]
female_cols = FEATURES_BY_SEX["Female"] + ["BodyFat"]

processed_male = datasets["Male"][male_cols]
processed_female = datasets["Female"][female_cols]

processed_male.to_csv("../data/processed/processed_male.csv", index=False)
processed_female.to_csv("../data/processed/processed_female.csv", index=False)
```
